```python
# --- Environment setup (works locally and in Colab) ---
import os, sys, subprocess
from pathlib import Path

IN_COLAB = "google.colab" in sys.modules
REPO = "dtenreiro/nb"

if IN_COLAB:
    if not Path("/content/nb").exists():
        subprocess.run(["git", "clone", "-q", f"https://github.com/{REPO}.git", "/content/nb"], check=True)
    ROOT = Path("/content/nb")
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(ROOT / "requirements.txt")], check=True)
else:
    ROOT = Path.cwd().resolve()
    while not (ROOT / "requirements.txt").exists() and ROOT != ROOT.parent:
        ROOT = ROOT.parent

DATA_DIR = ROOT / "data"
RAW = DATA_DIR / "raw"
PROCESSED = DATA_DIR / "processed"
PROCESSED.mkdir(parents=True, exist_ok=True)

print(f"Colab: {IN_COLAB}\nROOT:  {ROOT}\nRAW:   {RAW}")
```
